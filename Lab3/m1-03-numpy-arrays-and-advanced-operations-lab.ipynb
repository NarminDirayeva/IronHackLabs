# Task-3

```python
rng =np.random.default_rng(seed=0)
readings=rng.normal(50,10,size=(360,4))
```

```python
outlier_mask=(readings<20)|(readings>80)
```

```python
outlier_count=np.sum(outlier_mask)
outlier_count
```

```python
total_count=readings.size
total_count
```

```python
outlier_percent=(outlier_count/total_count)*100
outlier_percent
```

```python
cleaned_readings=readings.copy()
```

```python
cleaned_readings[outlier_mask]=np.nan
cleaned_readings
```

```python
nan_count=np.sum(np.isnan(cleaned_readings))
nan_count
```

```python
print(nan_count==outlier_count)
```
