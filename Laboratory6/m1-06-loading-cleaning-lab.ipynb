# Task-2

```python
tickets["opened_at"]=pd.to_datetime(tickets["opened_at"])
```

```python
tickets
```

```python
tickets["category"]=tickets["category"].str.strip().str.lower()
tickets["priority"]=tickets["priority"].str.strip().str.lower()
```

```python
tickets["category"].unique()
```

```python
tickets["priority"].unique()
```

# Task-3

```python
tickets["resolution_minutes"]=pd.to_numeric(tickets["resolution_minutes"],errors="coerce")
```

```python
tickets["resolution_minutes"].isna().sum()
```

# Task-4

```python
tickets_clean=tickets.dropna(subset=["resolution_minutes"])
tickets_clean
```

```python
(tickets_clean["resolution_minutes"]>=0).all()
```

```python
print((tickets_clean["category"] == tickets_clean["category"].str.strip().str.lower()).all())
```

```python
print((tickets_clean["priority"] == tickets_clean["priority"].str.strip().str.lower()).all())
```

```python
print("Before cleaning: ", len(tickets))
print("After cleaning: ", len(tickets_clean))
```

```python
tickets_clean["resolution_minutes"].max()  
```

# Task-5

```python
total_records = len(tickets)
clean_records = len(tickets_clean)
missing_resolution = tickets["resolution_minutes"].isna().sum()
```

```python
summary={
    "total_records":total_records,
    "clean_records": clean_records,
    "missing_resolution":missing_resolution,
    "avg_resolution": tickets_clean["resolution_minutes"].mean() ,
    "max_resolution": tickets_clean["resolution_minutes"].max()                  
}
```

```python
print(f"""
total_records: {summary['total_records']}
clean_records: {summary['clean_records']}
missing_resolution: {summary['missing_resolution']}
avg_resolution: {summary['avg_resolution']}
max_resolution: {summary['max_resolution']}
""")
```
