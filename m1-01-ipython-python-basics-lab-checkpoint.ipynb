# Task-4

```python
user_events = [
    {"user_id": 1, "event_type": "session_start", "duration_seconds": 75},
    {"user_id": 2, "event_type": "stream_video", "duration_seconds": 540},
    {"user_id": 3, "event_type": "send_message", "duration_seconds": 30},
    {"user_id": 4, "event_type": "play_audio", "duration_seconds": 195},
    {"user_id": 5, "event_type": "read_content", "duration_seconds": 260},
    {"user_id": 6, "event_type": "play_match", "duration_seconds": -300},
    {"user_id": 7, "event_type": "session_end", "duration_seconds": 40},
    {"user_id": 8, "event_type": "watch_clip", "duration_seconds": 390},
    {"user_id": 9, "event_type": "listen_show", "duration_seconds": "two seconds"},
    {"user_id": 10, "event_type": "edit_profile", "duration_seconds": 55},
    {"user_id": 11, "event_type": "run_search", "duration_seconds": 25},
    {"user_id": 12, "event_type": "play_audio", "duration_seconds": 230}
]
cleaned_events=[]
for event in user_events:
    try:
         duration=float(event["duration_seconds"])
         if duration>=0:
             cleaned=event.copy()
             cleaned["duration_minutes"]=duration/60
             cleaned_events.append(cleaned)
   
    except:
        pass    
```

```python
expected=10

print("Cleaned events count:", len(cleaned_events))
print("Expected count:", expected)
```

```python
if len(cleaned_events)==expected:
    print("Counted correctly")
else:
    print("Didn't count correctly")
```

```python
keys=True
for event in cleaned_events:
    if("user_id" not in event and "event_type" not  in event and "duration_seconds" not in event and "duration_minutes" not in event):
        keys=False
if keys:
    print("Cleaned record contains all required keys ")
else:
    print("Cleaned record doesn't contain all required keys ")
```

# Task-5

```python
event_counts={}
for event in cleaned_events:
    etype=event["event_type"]
    if etype in event_counts:
       event_counts[etype]+=1
    else:
        event_counts[etype]=1
```

```python
total_mins={}
total_counts={}
for event in cleaned_events:
    etype=event["event_type"]
    mins=event["duration_minutes"]
    if etype in total_mins:
        total_mins[etype]+=mins
        total_counts[etype]+=1
    else:
        total_mins[etype]=mins
        total_counts[etype]=1
avg_mins={}
for et in total_mins:
    avg_mins[et]=total_mins[et]/total_counts[et]
```

```python
users=set()
for event in cleaned_events:
    users.add(event["user_id"])
unique_users=len(users)
```

```python
sum_of_counts=0
for count in event_counts.values():
    sum_of_counts+=count
if sum_of_counts==len(cleaned_events):
    print("Validation successful")
else:
    print("Validation unsuccessful")
    
```

```python
csv_lines=[]
csv_lines.append("metric,key,value")

```

```python
for etype in event_counts:
    line="count,"+etype+","+str(event_counts[etype])
    csv_lines.append(line)
```

```python
for etype in avg_mins:
    line="average,"+etype+","+str(avg_mins[etype])
    csv_lines.append(line)
```

```python
csv_lines.append("unique_users,all," + str(unique_users))
```

```python
csv_string = "\n".join(csv_lines)
```

```python
print("\nFinal CSV:\n")
print(csv_string)
```
